```python
import heapq
from typing import List

class Solution:
    def smallestChair(self, times: List[List[int]], targetFriend: int) -> int:
        n = len(times)
        for i in range(n):
            times[i].append(i)

        times.sort()
        
        idle = list(range(n))
        heapq.heapify(idle)
        
        
        busy = []

        for arrival, leaving, friend in times:
            while busy and busy[0][0] <= arrival:
                free_time, chair = heapq.heappop(busy)
                heapq.heappush(idle, chair)
            
            chair = heapq.heappop(idle)
            
            if friend == targetFriend:
                return chair
            
            heapq.heappush(busy, (leaving, chair))
        
        return -1
```

```python
times = [[6, 8], [2, 3], [4, 6]]
targetFriend = 1
sol = Solution()
print(sol.smallestChair(times, targetFriend))
```
