Merge Intervals
Level: Medium
Time: 30 min

Problem Description
Explore how to merge overlapping intervals by evaluating their start and end times. This lesson helps you understand the problem constraints and develop efficient solutions to return non-overlapping intervals. Gain practice in analyzing interval problems commonly encountered in coding interviews.

Statement
We are given an array of closed intervals called intervals, where each interval has a start time and an end time and is represented as intervals[i] = [starti, endi].

Your task is to merge all the overlapping intervals and return an array of the resulting non-overlapping intervals that cover all the intervals in the input.

Constraints

1 ≤ intervals.length ≤ 10³

intervals[i].length == 2

0 ≤ starti ≤ endi ≤ 10⁴


## Solution Approach
Key Insights
Closed Intervals: [start, end] includes both endpoints
Overlap Condition: Two intervals [s1,e1] and [s2,e2] overlap if s1 ≤ e2 AND s2 ≤ e1
Merge Strategy: Always take min(start) and max(end) of overlapping intervals
Sort First: Sort by start time to process intervals chronologically
Algorithm (O(n log n))


1. Sort intervals by start time
2. Initialize result with first interval
3. For each subsequent interval:
   - If it overlaps with last result interval:
     → Extend last interval: end = max(current_end, interval_end)
   - Else:
     → Add as new interval to result
4. Return result


## Example Walkthrough

Input: intervals = [[1,3],[2,6],[8,10],[15,18]]

Step 1: Sort → [[1,3],[2,6],[8,10],[15,18]]

Step 2: Process:

[1,3] → result = [[1,3]]

[2,6] overlaps (2≤3) → [1,6]

[8,10] no overlap (8>6) → [[1,6],[8,10]]

[15,18] no overlap → [[1,6],[8,10],[15,18]]

Output: [[1,6],[8,10],[15,18]]

```python
def merge(intervals):
    if not intervals:
        return []
    
    # Sort by start time
    intervals.sort(key=lambda x: x[0])
    
    result = [intervals[0]]
    
    for current in intervals[1:]:
        last = result[-1]
        
        # Overlap: extend last interval
        if current[0] <= last[1]:
            last[1] = max(last[1], current[1])
        # No overlap: add new interval
        else:
            result.append(current)
    
    return result
```
