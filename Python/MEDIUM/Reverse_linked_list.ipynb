# Reverse Linked List II

## Problem Statement

Given a singly linked list with `n` nodes and two positions, `left` and `right`, reverse the nodes of the list from position `left` to position `right`.

Return the modified list.

The reversal must be done **in-place**.

---

## Constraints

```text
1 <= n <= 500
-5000 <= node.value <= 5000
1 <= left <= right <= n
