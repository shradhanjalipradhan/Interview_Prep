```python
class ListNode:
    def __init__(self, val=0, next=None):
        self.val = val
        self.next = next

def build_list(values):
    dummy = ListNode(0)
    curr = dummy
    for v in values:
        curr.next = ListNode(v)
        curr = curr.next
    return dummy.next

head = build_list([1, 2, 3, 4, 5])
dummy = ListNode(0)
dummy.next = head
prev = dummy
```
