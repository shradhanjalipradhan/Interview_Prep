## Fruit Into Baskets

### Problem Description
Explore how to apply the sliding window pattern to solve the Fruit Into Baskets problem. Learn to maximize fruit collection from a row with two baskets following specific rules. Understand problem constraints and develop an optimized approach for efficient solutions.

### Statement
You have a row of fruits represented by an integer array fruits, where fruits[i] is the type of fruit the iᵗʰ tree produces.

### Rules:

1. You have only two baskets, each holding unlimited quantity of ONE type of fruit
2. Start from any tree, collect exactly one fruit per tree moving right
3. Stop when encountering a fruit type that doesn't fit either basket
4. Goal: Return the maximum number of fruits you can collect.

### Constraints

1 ≤ fruits.length ≤ 10³

0 ≤ fruits[i] < fruits.length
### Solution Approach
Key Insights
1. Two Basket Limit = At most 2 distinct fruit types in window
2. Sliding Window: Expand right, shrink left when > 2 types
3. Track Basket Contents: Use hashmap to count frequencies
4. Maximize Window: Longest valid subarray with ≤ 2 distinct types
5. Algorithm (O(n))


1. Initialize left=0, max_fruits=0, basket = {}
2. For each right position:
   a. Add fruits[right] to basket
   b. While > 2 distinct fruits:
      → Remove fruits[left] from basket
      → left += 1
   c. Update max_fruits = max(max_fruits, right - left + 1)
3. Return max_fruits

```python
def totalFruit(fruits):
    basket = {}
    left = 0
    max_fruits = 0
    
    for right in range(len(fruits)):
        # Add fruit to basket
        fruit = fruits[right]
        basket[fruit] = basket.get(fruit, 0) + 1
        
        # Shrink window if > 2 types
        while len(basket) > 2:
            left_fruit = fruits[left]
            basket[left_fruit] -= 1
            if basket[left_fruit] == 0:
                del basket[left_fruit]
            left += 1
        
        # Update maximum
        max_fruits = max(max_fruits, right - left + 1)
    
    return max_fruits
```
