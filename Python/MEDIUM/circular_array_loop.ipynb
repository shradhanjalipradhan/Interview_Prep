    def circularArrayLoop(self, nums):
        n = len(nums)

        def next_index(i):
            return (i + nums[i]) % n

        for start in range(n):
            if nums[start] == 0:
                continue

            slow = start
            fast = next_index(start)

            while nums[slow] * nums[fast] > 0 and nums[slow] * nums[next_index(fast)] > 0:
                if slow == fast:
                    if slow != next_index(slow):
                        return True
                    break
                slow = next_index(slow)
                fast = next_index(next_index(fast))

            i = start
            sign = nums[start]
            while nums[i] * sign > 0:
                nxt = next_index(i)
                nums[i] = 0
                i = nxt

        return False
