Frequency of the Most Frequent Element

Level: Medium

Time: 30 min

Problem Description

Explore how to apply the sliding window technique to maximize the frequency of an element in an integer array by incrementing values within a set limit. Understand how to efficiently solve problems involving frequency calculation and constraints to improve your coding interview skills.

Statement

You are given an integer array, nums, and an integer k, representing the maximum number of operations you can perform. In each operation, you may select any index in nums and increment its value by 1.

Your task is to determine the maximum possible frequency of a single element in the final array after performing at most k operations.

You can choose to increase any elements in a way that results in one particular element appearing as often as possible (within k operations).

Example: nums = [2, 2, 3], k = 4

→ Increment first 2→3 (1 op), second 2→3 (1 op)

→ Result: [3, 3, 3], frequency = 3

### Constraints

1 ≤ nums.length ≤ 10³

1 ≤ nums[i] ≤ 10³

1 ≤ k ≤ 10³

### Solution Approach
Key Insights
1. Target Element: For each possible target value, find max window where all elements ≤ target
2. Operations Needed: To make nums[j] equal target: target - nums[j] operations
3. Sliding Window: Expand right, shrink left when operations exceed k
4. Max Frequency: Track maximum window size

Algorithm (O(n²) → Optimized O(n log n))


1. Sort nums (enables sliding window)
2. For each possible right endpoint:
   a. Expand window [left, right]
   b. Calculate operations needed: sum(target - nums[j]) for j in window
   c. Shrink left if operations > k
   d. Update max_freq = max(max_freq, right - left + 1)
3. Return max_freq

```python
def maxFrequency(nums, k):
    nums.sort()
    left = 0
    max_freq = 0
    operations = 0
    
    for right in range(len(nums)):
        # Add right element (target = nums[right])
        operations += nums[right] - nums[left]
        
        # Shrink window if operations exceed k
        while operations > k:
            operations -= nums[right] - nums[left]
            left += 1
        
        # Update max frequency
        max_freq = max(max_freq, right - left + 1)
    
    return max_freq
```
